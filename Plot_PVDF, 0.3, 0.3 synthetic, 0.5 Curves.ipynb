## Plotting PVDF, PGO-O.3%, PGO-0.3% Synthetic, PGO-0.5%

```python
# =====================================
# 1. Data Read
# =====================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


def read_data(filename, encoding):
    return pd.read_csv(
        filename,
        sep=r"\s+",
        header=None,
        names=["x", "y"],
        encoding=encoding,
        engine="python"
    )


df1 = read_data("Pure Polymer.txt", "utf-8")
df2 = read_data("PGO-0.3%.txt", "utf-8")
df3 = read_data("PGO-0.3_synthetic_noise_pattern_from_0.5.txt", "utf-8")
df4 = read_data("PGO-0.5%.txt", "utf-16")


# =====================================
# 2. Create Figure
# =====================================

fig, ax = plt.subplots(figsize=(10, 7))


# =====================================
# 3. Plot XRD Curves
# =====================================

ax.plot(
    df1["x"],
    df1["y"],
    label="PVDF",
    linewidth=2
)

ax.plot(
    df2["x"],
    df2["y"] + 1200,
    label="PGO-0.3%",
    linewidth=2
)

ax.plot(
    df3["x"],
    df3["y"] + 2600,
    label="PGO-0.3%_synthetic",
    linewidth=2
)

ax.plot(
    df4["x"],
    df4["y"] + 4000,
    label="PGO-0.5%",
    linewidth=2
)


# =====================================
# 4. XRD Reference Positions
#    Based on the main article
# =====================================

go_peak = 9.4

alpha_1 = 18.8
beta = 20.5

alpha_2 = 36.5
alpha_3 = 40.1


# =====================================
# 5. Reference Line Color
# =====================================

line_color = "mediumseagreen"


# =====================================
# 6. Vertical Reference Lines
# =====================================

# -----------------------------
# GO peak
# -----------------------------

ax.vlines(
    go_peak,
    ymin=0,
    ymax=6500,
    color=line_color,
    linewidth=2,
    alpha=0.65
)


# -----------------------------
# Alpha 18.8°
# -----------------------------

ax.vlines(
    alpha_1,
    ymin=0,
    ymax=7500,
    color=line_color,
    linewidth=2,
    alpha=0.65
)


# -----------------------------
# Beta 20.5°
# -----------------------------

ax.vlines(
    beta,
    ymin=0,
    ymax=9000,
    color=line_color,
    linewidth=2,
    alpha=0.65
)


# -----------------------------
# Alpha 36.5°
# -----------------------------

ax.vlines(
    alpha_2,
    ymin=0,
    ymax=6100,
    color=line_color,
    linewidth=2,
    alpha=0.65
)


# -----------------------------
# Alpha 40.1°
# -----------------------------

ax.vlines(
    alpha_3,
    ymin=0,
    ymax=6100,
    color=line_color,
    linewidth=2,
    alpha=0.65
)


# =====================================
# 7. Peak Labels
# =====================================

# -----------------------------
# GO peak
# -----------------------------

ax.text(
    go_peak,
    6700,
    "GO peak",
    ha="center",
    va="bottom",
    fontsize=15,
    fontweight="bold"
)


# -----------------------------
# Alpha at 18.8°
# -----------------------------

ax.text(
    alpha_1 - 0.4,
    7700,
    r"$\alpha$",
    ha="center",
    va="bottom",
    fontsize=25,
    fontweight="bold"
)


# -----------------------------
# Beta at 20.5°
# -----------------------------

ax.text(
    beta + 0.5,
    9100,
    r"$\beta$",
    ha="center",
    va="bottom",
    fontsize=25,
    fontweight="bold"
)


# -----------------------------
# Alpha at 36.5°
# -----------------------------

ax.text(
    alpha_2,
    6500,
    r"$\alpha$",
    ha="center",
    va="bottom",
    fontsize=25,
    fontweight="bold"
)


# -----------------------------
# Alpha at 40.1°
# -----------------------------

ax.text(
    alpha_3,
    6500,
    r"$\alpha$",
    ha="center",
    va="bottom",
    fontsize=25,
    fontweight="bold"
)


# =====================================
# 8. Axis Labels
# =====================================

ax.set_xlabel(
    r"2$\theta$ (degree)",
    fontsize=19
)

ax.set_ylabel(
    "Intensity (a.u.)",
    fontsize=19
)


# =====================================
# 9. Tick Font Size
# =====================================

ax.tick_params(
    axis="x",
    labelsize=15
)

ax.tick_params(
    axis="y",
    left=False,
    labelleft=False
)


# =====================================
# 10. Legend
# =====================================

ax.legend(
    loc="upper left",
    bbox_to_anchor=(0.02, 0.99),
    fontsize=17,
    frameon=False
)


# =====================================
# 11. Axis Limits
# =====================================

ax.set_xlim(5, 80)
ax.set_ylim(0, 14000)


# =====================================
# 12. Remove Grid
# =====================================

ax.grid(False)


# =====================================
# 13. Layout
# =====================================

plt.tight_layout()

plt.show()
```
